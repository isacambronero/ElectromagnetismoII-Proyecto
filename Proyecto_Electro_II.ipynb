## 1. Enunciado 

Considere una onda electromagnética plana, monocromática y linealmente polarizada que se propaga en la dirección $+z$ dentro de un material birrefringente.
La polarización inicial forma un ángulo $\theta$ con el eje $x$. Debido a la birrefringencia, las componentes $x$ y $y$ del campo eléctrico experimentan índices de refracción diferentes, $n_x$ y $n_y$, por lo que acumulan una diferencia de fase al propagarse.
Cree un programa que permita visualizar cómo cambia la polarización de la onda al variar:
- el ángulo inicial $\theta$,
- el espesor $L$,
- la birrefringencia $\Delta n=n_y-n_x$,
- y la longitud de onda $\lambda_0$.

Se deben mostrar las componentes $E_x$ y $E_y$, la diferencia de fase acumulada y la trayectoria de la punta del campo eléctrico para identificar polarización lineal, circular o elíptica.


## 2. Solución analítica 
La onda incidente está linealmente polarizada en el plano transversal $xy$. Su amplitud puede descomponerse como

$$
\mathbf{E}_0
=
E_0\cos\theta\,\hat{\mathbf{x}}
+
E_0\sin\theta\,\hat{\mathbf{y}}.
$$

En un medio birrefringente, cada componente experimenta un índice de refracción distinto. Para una longitud de onda en vacío $\lambda_0$,

$$
k_x=\frac{2\pi n_x}{\lambda_0},
\qquad
k_y=\frac{2\pi n_y}{\lambda_0}.
$$

Por lo tanto, las soluciones que utilizaremos serían:

$$
E_x(z,t)
=
E_0\cos\theta\,
\cos(k_xz-\omega t),
$$

$$
E_y(z,t)
=
E_0\sin\theta\,
\cos(k_yz-\omega t).
$$

La diferencia de fase entre ambas componentes después de propagarse una distancia $z$ es

$$
\Delta\phi(z)
=
(k_y-k_x)z.
$$

Como

$$
k_y-k_x
=
\frac{2\pi}{\lambda_0}(n_y-n_x),
$$

se obtiene la relación:

$$
\boxed{
\Delta\phi(z)
=
\frac{2\pi\Delta n}{\lambda_0}\,z
}
$$

donde

$$
\Delta n=n_y-n_x.
$$

A la salida del material, $z=L$,

$$
\boxed{
\Delta\phi_L
=
\frac{2\pi\Delta n L}{\lambda_0}
}.
$$

### Algunos casos importantes

- Si una de las componentes tiene amplitud nula, la onda es linealmente polarizada.
- Si $\Delta\phi_L=m\pi$, las componentes están en fase o antifase y la polarización es lineal.
- Si las amplitudes son iguales, lo cual ocurre para $\theta=45^\circ$, y

$$
\Delta\phi_L
=
\frac{\pi}{2}+m\pi,
$$

la polarización es circular.

- En el caso general, la polarización es elíptica.

### Suposiciones 
Para mantener este proyecto basado en soluciones analíticas, vamos a considerar: 
- la onda es plana y monocromática
- se propaga en dirección $+z$
- los ejes $x$ y $y$ representan dos direcciones principales del material birrefringente 
- el material es lineal, homogéneo y no absorbente
- no se van a incluir reflexiones en las superficies de entrada y salida
- al variar $\lambda_0$, los valores de $n_x$ y $\Delta n$ se mantienen independientes de la longitud de onda, por lo que no se estudia dispersión real del material.


### 3. Librerías 
Se utilizan NumPy para evaluar las expresiones analíticas, Matplotlib para las gráficas e ipywidgets para los controles interactivos. Los sliders únicamente modifican los parámetros que se sustituyen en las ecuaciones anteriores.

```python
import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets
from IPython.display import display
```

### 4. Clasificación de polarización

Aquí crearemos una función para utilizar la amplitud relativa de las componentes y la diferencia de fase a la salida del material. 

Tendremos que es:
- lineal si una componente es prácticamente nula o si $\sin(\Delta \phi) \approx 0$
- circular si ambas tienen igual amplitud y el desfase es aproximadamente $\pi/2 + m\pi$
- elíptica en otros casos 


```python
def clasificar_polarizacion(theta, delta_phi):
    Ax = abs(np.cos(theta))
    Ay = abs(np.sin(theta))
    delta = np.mod(delta_phi, 2 * np.pi)
    
    # Si el seno de la fase es cero (o múltiplo de pi), es lineal
    if np.isclose(np.sin(delta), 0, atol=0.05):
        return "Polarización lineal"
    # Si las amplitudes son iguales y la fase es pi/2 o 3pi/2, es circular
    elif (
        np.isclose(Ax, Ay, atol=0.05)
        and np.isclose(abs(np.sin(delta)), 1, atol=0.05)
    ):
        return "Polarización circular"
    else:
        # Cualquier otro caso intermedio es elíptica
        return "Polarización elíptica"
```

### 5. Parámetros iniciales

Se toma inicialmente una onda con longitud de onda en vacío $\lambda_0 = 550 nm$ y un índice de refracción $n_x = 1.50$. 

La birrefringencia inicial se toma como $\Delta n = 0.05$, por lo que $n_y = 1.55$

También al inicio $\theta = 45^\circ$ y un espesor de $L = 4.0 \mu m$

Se tendrán controles que nos van a permitir variar de manera interactiva los siguientes parámetros: 
- $\theta$: orientación de la polarización lineal 
- $L$: distancia recorrida dentro del material 
- $\Delta n$: magnitud de birrefringencia 
- $\lambda_0$ longitud de onda en el vacío 
- $\omega t$: fase temporal usada para observar un instande de $E_x(z,t)$ y $E_y(z,t)$

```python
# Parámetros base
E0 = 1.0
nx_base = 1.50

lambda0_ini_nm = 550.0
delta_n_ini = 0.05
theta_ini_deg = 45.0

L_ini_um = 4.0

# Controles
theta_slider = widgets.FloatSlider(
    value=theta_ini_deg,
    min=0.0,
    max=90.0,
    step=1.0,
    description="θ [°]:",
    continuous_update=False,
    style={"description_width": "90px"}
)

L_slider = widgets.FloatSlider(
    value=L_ini_um,
    min=0.25,
    max=50.0,
    step=0.25,
    description="L [μm]:",
    continuous_update=False,
    style={"description_width": "90px"}
)

dn_slider = widgets.FloatSlider(
    value=delta_n_ini,
    min=0.001,
    max=0.10,
    step=0.001,
    description="Δn:",
    readout_format=".3f",
    continuous_update=False,
    style={"description_width": "90px"}
)

lambda_slider = widgets.FloatSlider(
    value=lambda0_ini_nm,
    min=400.0,
    max=700.0,
    step=5.0,
    description="λ₀ [nm]:",
    continuous_update=False,
    style={"description_width": "90px"}
)

fase_slider = widgets.FloatSlider(
    value=0.0,
    min=0.0,
    max=2*np.pi,
    step=np.pi/24,
    description="ωt [rad]:",
    readout_format=".2f",
    continuous_update=False,
    style={"description_width": "90px"}
)
```

### 6. Visualización 

La figura que vamos a ver de resultados tiene 4 secciones: 

- **Componentes $E_x(z,t)$ $E_y(z, t)$**

En esta figura se muestra un instante de las dos componente del campo mientras se propagan en el material. Con el slider se puede cambiar el instante observado. 

- **Estado de polarización a la salida del material**

En esta se grafica de manera paramétrica: $E_x(L,t)$ $E_y(L, t)$ en un periodo completo. La trayectoria de la punta del vector **E** permite identificar una línea, circunferencia o elipse. El punto marcado sobre la trayectoria corresponde al valor instantáneo elegido con el slider anterior. 

- **Desfase acumulado**

Se muestra
$$
\Delta\phi(z)=\frac{2\pi\Delta n}{\lambda_0}z,
$$
para observar cómo aumenta el desfase a medida que la onda avanza dentro del material.

- **Evolución de la polarización dentro del material**

Se muestran las curvas de polarización en varias posiciones:
$$
z=0,\quad
z=\frac{L}{4},\quad
z=\frac{L}{2},\quad
z=\frac{3L}{4},\quad
z=L.
$$
Este panel permite observar directamente cómo cambia la trayectoria del campo conforme se acumula el desfase.

```python
def visualizar(theta_deg, L_um, delta_n, lambda_nm, phase_t):
    theta = np.deg2rad(theta_deg)
    L = L_um * 1e-6
    lambda0 = lambda_nm * 1e-9

    nx = nx_base
    ny = nx + delta_n

    kx = 2*np.pi*nx/lambda0
    ky = 2*np.pi*ny/lambda0

    z = np.linspace(0, L, 600)
    fase = np.linspace(0, 2*np.pi, 500)

    # Campo dentro del material
    Ex = E0*np.cos(theta)*np.cos(kx*z - phase_t)
    Ey = E0*np.sin(theta)*np.cos(ky*z - phase_t)

    # Diferencia de fase acumulada
    delta_phi = (ky - kx)*z
    delta_phi_L = delta_phi[-1]

    # Estado de polarización a la salida
    Ex_out = E0*np.cos(theta)*np.cos(kx*L - fase)
    Ey_out = E0*np.sin(theta)*np.cos(ky*L - fase)

    Ex_now = E0*np.cos(theta)*np.cos(kx*L - phase_t)
    Ey_now = E0*np.sin(theta)*np.cos(ky*L - phase_t)

    tipo = clasificar_polarizacion(theta, delta_phi_L)

    fig, axes = plt.subplots(2, 2, figsize=(13, 9))
    ax1, ax2, ax3, ax4 = axes.flatten()

    # 1. Componentes del campo
    ax1.plot(z*1e6, Ex, label=r"$E_x$")
    ax1.plot(z*1e6, Ey, label=r"$E_y$")
    ax1.set_xlabel(r"$z\ [\mu m]$")
    ax1.set_ylabel(r"$E/E_0$")
    ax1.set_title("Componentes del campo")
    ax1.set_ylim(-1.1, 1.1)
    ax1.grid(True, alpha=0.3)
    ax1.legend()

    # 2. Polarización a la salida
    ax2.plot(Ex_out, Ey_out)
    ax2.plot(Ex_now, Ey_now, "o")
    ax2.set_xlabel(r"$E_x/E_0$")
    ax2.set_ylabel(r"$E_y/E_0$")
    ax2.set_title(
        f"{tipo}\n"
        + rf"$\Delta\phi_L/\pi={delta_phi_L/np.pi:.3f}$"
    )
    ax2.set_xlim(-1.1, 1.1)
    ax2.set_ylim(-1.1, 1.1)
    ax2.set_aspect("equal")
    ax2.grid(True, alpha=0.3)

    # 3. Desfase acumulado
    ax3.plot(z*1e6, delta_phi)
    ax3.set_xlabel(r"$z\ [\mu m]$")
    ax3.set_ylabel(r"$\Delta\phi$ [rad]")
    ax3.set_title("Diferencia de fase acumulada")
    ax3.grid(True, alpha=0.3)

    # 4. Evolución de la polarización
    for frac in [0, 0.25, 0.50, 0.75, 1.0]:
        zi = frac*L
        Ex_i = E0*np.cos(theta)*np.cos(kx*zi - fase)
        Ey_i = E0*np.sin(theta)*np.cos(ky*zi - fase)
        ax4.plot(Ex_i, Ey_i, label=rf"$z/L={frac:.2f}$")

    ax4.set_xlabel(r"$E_x/E_0$")
    ax4.set_ylabel(r"$E_y/E_0$")
    ax4.set_title("Evolución de la polarización")
    ax4.set_xlim(-1.1, 1.1)
    ax4.set_ylim(-1.1, 1.1)
    ax4.set_aspect("equal")
    ax4.grid(True, alpha=0.3)
    ax4.legend(fontsize=8)

    fig.suptitle(
        rf"$n_x={nx:.3f}$, $n_y={ny:.3f}$, "
        rf"$\lambda_0={lambda_nm:.0f}$ nm, "
        rf"$L={L_um:.2f}\,\mu m$",
        fontsize=14
    )

    plt.tight_layout()
    plt.show()

    print(f"Estado a la salida: {tipo}")
    print(f"Δφ(L) = {delta_phi_L:.5f} rad = {delta_phi_L/np.pi:.5f}π")
```

```python
controls_left = widgets.VBox([theta_slider, L_slider, dn_slider])
controls_right = widgets.VBox([lambda_slider, fase_slider])

ui = widgets.HBox([controls_left, controls_right])

out = widgets.interactive_output(
    visualizar,
    {
        "theta_deg": theta_slider,
        "L_um": L_slider,
        "delta_n": dn_slider,
        "lambda_nm": lambda_slider,
        "phase_t": fase_slider,
    },
)

display(ui, out)
```
